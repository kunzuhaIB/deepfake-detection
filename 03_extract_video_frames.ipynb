```python
import cv2
import os
import numpy as np

test_videos_set = set(metadata_df_final_v1[metadata_df_final_v1['split'] == 'test']['video_id'])
val_videos_set = set(metadata_df_final_v1[metadata_df_final_v1['split'] == 'val']['video_id'])

# OpenCV Haar cascade for lightweight CPU face detection
FACE_CASCADE_PATH = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(FACE_CASCADE_PATH)


def get_output_folder(video_name: str, label: str, base_out_dir: str) -> str:
    if video_name in test_videos_set:
        return os.path.join(base_out_dir, 'test', label)
    elif video_name in val_videos_set:
        return os.path.join(base_out_dir, 'val', label)
    else:
        return os.path.join(base_out_dir, 'train', label)


def crop_largest_face(
    frame: np.ndarray,
    scale_factor: float = 1.1,
    min_neighbors: int = 5,
    min_size: tuple = (60, 60),
    padding_ratio: float = 0.25
):
    """
    Detect the largest face and return a padded crop.
    Returns None if no face is found.
    """
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=scale_factor,
        minNeighbors=min_neighbors,
        minSize=min_size
    )

    if len(faces) == 0:
        return None

    # Choose the largest detected face
    x, y, w, h = max(faces, key=lambda b: b[2] * b[3])

    pad_w = int(w * padding_ratio)
    pad_h = int(h * padding_ratio)

    x1 = max(0, x - pad_w)
    y1 = max(0, y - pad_h)
    x2 = min(frame.shape[1], x + w + pad_w)
    y2 = min(frame.shape[0], y + h + pad_h)

    face_crop = frame[y1:y2, x1:x2]
    if face_crop.size == 0:
        return None

    return face_crop


def is_near_duplicate(
    prev_crop_gray: np.ndarray,
    curr_crop_gray: np.ndarray,
    diff_threshold: int = 12,
    change_ratio_threshold: float = 0.015
) -> bool:
    """
    Returns True if crops are too similar.
    """
    if prev_crop_gray is None:
        return False

    # Resize both to a fixed small size for cheap comparison
    target_size = (128, 128)
    a = cv2.resize(prev_crop_gray, target_size)
    b = cv2.resize(curr_crop_gray, target_size)

    diff = cv2.absdiff(a, b)
    _, thresh = cv2.threshold(diff, diff_threshold, 255, cv2.THRESH_BINARY)
    change_ratio = np.count_nonzero(thresh) / thresh.size

    return change_ratio < change_ratio_threshold


def extract_xception_frames(
    source_video_path: str,
    video_name: str,
    output_folder: str,
    frame_interval: int = 10,
    max_frames_per_video: int = 15,
    output_size: tuple = (299, 299),
    duplicate_filter: bool = True,
    diff_threshold: int = 12,
    change_ratio_threshold: float = 0.015,
    save_center_crop_if_no_face: bool = False
) -> list:
    """
    Extract face crops suitable for Xception training.

    Strategy:
    - sample every Nth frame
    - detect largest face
    - skip near-duplicate face crops
    - cap number of saved crops per video
    - resize to Xception size (299x299)
    """
    os.makedirs(output_folder, exist_ok=True)
    saved_paths = []

    cap = cv2.VideoCapture(source_video_path)
    if not cap.isOpened():
        print(f"Could not open video: {source_video_path}")
        return saved_paths

    frame_count = 0
    saved_count = 0
    prev_crop_gray = None

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_count % frame_interval != 0:
            frame_count += 1
            continue

        if saved_count >= max_frames_per_video:
            break

        face_crop = crop_largest_face(frame)

        if face_crop is None and save_center_crop_if_no_face:
            h, w = frame.shape[:2]
            side = min(h, w)
            x1 = (w - side) // 2
            y1 = (h - side) // 2
            face_crop = frame[y1:y1 + side, x1:x1 + side]

        if face_crop is None:
            frame_count += 1
            continue

        crop_gray = cv2.cvtColor(face_crop, cv2.COLOR_BGR2GRAY)

        if duplicate_filter and is_near_duplicate(
            prev_crop_gray,
            crop_gray,
            diff_threshold=diff_threshold,
            change_ratio_threshold=change_ratio_threshold
        ):
            frame_count += 1
            continue

        resized = cv2.resize(face_crop, output_size, interpolation=cv2.INTER_AREA)

        frame_filename = f"{video_name}_face_{saved_count:03d}.jpg"
        frame_path = os.path.join(output_folder, frame_filename)
        cv2.imwrite(frame_path, resized)

        saved_paths.append(frame_path)
        saved_count += 1
        prev_crop_gray = crop_gray
        frame_count += 1

    cap.release()
    print(f"[{video_name}] saved {saved_count} face crops")
    return saved_paths


base_data_dir = '/Users/kunzuhaib/work/mycode/student/trimester_02/project/data'
base_out_dir = '/Users/kunzuhaib/work/mycode/student/trimester_02/project/data_optimal_frames'


for directory in ['Celeb-synthesis', 'YouTube-real', 'Celeb-real']:
    source_folder = os.path.join(base_data_dir, directory)
    label = 'real' if 'real' in directory else 'fake'

    for video in os.listdir(source_folder):
        if video.startswith('.'):
            continue

        source_video_path = os.path.join(source_folder, video)
        output_folder = get_output_folder(video, label, base_out_dir)

        extract_xception_frames(
            source_video_path=source_video_path,
            video_name=video,
            output_folder=output_folder,
            frame_interval=10,            # every 10th frame
            max_frames_per_video=15,      # cap per-video samples
            output_size=(299, 299),       # Xception default size
            duplicate_filter=True,
            diff_threshold=12,
            change_ratio_threshold=0.015,
            save_center_crop_if_no_face=False
        )
```

```python
import tensorflow as tf
from tensorflow.keras import layers
from tensorflow.keras.applications import Xception

base_model = Xception(
    weights='imagenet',       # Use weights pre-trained on the massive ImageNet dataset
    include_top=False,        # Drop the original 1000-class classifier
    input_shape=(299, 299, 3) # The shape we verified in your TensorSpec
)

# 2. Freeze the base model so the CPU doesn't try to update 20+ million weights
base_model.trainable = False 

# 3. Construct the new model architecture
inputs = tf.keras.Input(shape=(299, 299, 3))

# Pass inputs through the frozen base model
# training=False ensures BatchNormalization layers stay in inference mode
x = base_model(inputs, training=False) 

# Convert the 2D feature maps from Xception into a 1D vector
x = layers.GlobalAveragePooling2D()(x)

# Add Dropout to prevent the new layer from overfitting to your training data
x = layers.Dropout(0.5)(x)

# Final classification layer (1 neuron with sigmoid activation for binary Real/Fake output)
outputs = layers.Dense(1, activation='sigmoid')(x)

# Combine into a final Keras model
model = tf.keras.Model(inputs, outputs)

# 4. Compile the model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
print("Model built and compiled successfully!")
```

```python
import numpy as np
import tensorflow as tf
import cv2
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.xception import preprocess_input

def make_gradcam_heatmap(img_array, model, last_conv_layer_name="block14_sepconv2_act"):
    """
    Calculates the Grad-CAM heatmap for a given image array and model.
    """
    # 1. Extract the nested Xception base model and your custom top layers
    base_model = model.get_layer('xception')
    gap_layer = model.get_layer('global_average_pooling2d')
    dense_layer = model.get_layer('dense')
    # Note: We skip the dropout layer because it is inactive during inference

    # 2. Create a sub-model that outputs the last convolutional layer from Xception
    #    AND the final output of the Xception base model
    cam_base_model = tf.keras.Model(
        inputs=base_model.input,
        outputs=[base_model.get_layer(last_conv_layer_name).output, base_model.output]
    )

    # 3. Record operations for automatic differentiation
    with tf.GradientTape() as tape:
        # Pass image through the base model
        last_conv_layer_output, base_output = cam_base_model(img_array, training=False)
        
        # Manually pass the output through your custom top layers
        x = gap_layer(base_output)
        preds = dense_layer(x)
        
        # We are doing binary classification, so our target is just the single output node
        class_channel = preds[0][0]

    # 4. Calculate gradients of the prediction with respect to the last conv layer
    grads = tape.gradient(class_channel, last_conv_layer_output)

    # 5. Pool the gradients over the spatial dimensions
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    # 6. Multiply each channel in the feature map by "how important this channel is"
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    # 7. Normalize the heatmap between 0 and 1 for visualization
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

def display_gradcam(img_path, heatmap, alpha=0.4):
    """
    Overlays the heatmap onto the original image and displays them side-by-side.
    """
    # Load and format the original image for plotting
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (299, 299))

    # Resize the heatmap to match the original image size
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))

    # Convert the heatmap to RGB colors (JET colormap puts red on high values, blue on low)
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB) 

    # Superimpose the heatmap onto the original image
    superimposed_img = cv2.addWeighted(heatmap, alpha, img, 1 - alpha, 0)

    # Plotting
    fig, ax = plt.subplots(1, 2, figsize=(12, 6))
    ax[0].imshow(img)
    ax[0].set_title("Original Cropped Face")
    ax[0].axis('off')
    
    ax[1].imshow(superimposed_img)
    ax[1].set_title("Grad-CAM Heatmap")
    ax[1].axis('off')
    
    plt.tight_layout()
    plt.show()

# ==========================================
# --- How to Execute and Test an Image ---
# ==========================================

# 1. Point this to a specific test image (try to find one it got wrong, and one it got right!)
test_image_path = "/Users/kunzuhaib/work/mycode/student/trimester_02/project/data_optimal_frames/train/fake/id0_id1_0000.mp4_face_004.jpg"

# 2. Preprocess the image exactly like the pipeline did during training
img = load_img(test_image_path, target_size=(299, 299))
img_array = img_to_array(img)
img_array = np.expand_dims(img_array, axis=0) # Add the batch dimension (1, 299, 299, 3)
img_array = preprocess_input(img_array)       # Scale pixels to -1 to 1

# 3. Generate the Heatmap
heatmap = make_gradcam_heatmap(img_array, model)

# 4. Print the Prediction and Display
prediction = model.predict(img_array)[0][0]
print(f"Model Prediction Score: {prediction:.4f}")
print("(Scores closer to 1.0 indicate 'Fake', closer to 0.0 indicate 'Real')")

display_gradcam(test_image_path, heatmap)
```
